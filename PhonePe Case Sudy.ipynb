# Exploratory Data Analysis

```python
print("Total no of Transactions and Total Transaction amount for each state over the years:")
summary = state_txn.groupby(['State','Year'])[['Transactions','Amount (INR)']].sum().reset_index()
print(summary.to_string(index=False))
```

```python
State_transactions = state_txn.groupby('State')['Transactions'].sum().sort_values(ascending=False)
top_5 = State_transactions.head().reset_index()
bottom_5 = State_transactions.tail().reset_index()
print("Top 5 States with highest transaction volumes:")
print(top_5)
print("\n\nTop 5 States with lowest transaction volumes:")
print(bottom_5)
```

```python
print("Most Frequent Transaction Type for each State and Quarter:")
most_frequent = (
    state_txnSplit.loc[state_txnSplit.groupby(['State', 'Year', 'Quarter'])['Transactions'].idxmax()]
    [['State', 'Year', 'Quarter', 'Transaction Type', 'Transactions']]
    .sort_values(['State', 'Year', 'Quarter'])
    .reset_index(drop=True)
)
print(most_frequent.to_string(index=False))
```

```python
print("Device brand with the highest number of users in each state:")
users = state_dev_data.groupby(['State','Brand'])['Registered Users'].sum().reset_index()
top_brands = users.loc[users.groupby('State')['Registered Users'].idxmax()].sort_values('State').reset_index(drop=True)
print(top_brands.to_string(index=False))
```

```python
print("District with the highest population in each State:")
districts = dist_dem.loc[dist_dem.groupby('State')['Population'].idxmax()].reset_index()
print(districts[['State','District','Population']])
```

```python
plt.figure(figsize=(14,8),dpi=100)
plot = sns.barplot(x='State',y='Population',data=districts,hue='District',palette='Paired')
plt.xticks(rotation=90)
plt.title('Top District by Population per State', fontsize=14, fontweight='bold')
plt.legend(
    loc='center left',       
    bbox_to_anchor=(1.05, 0.5),
    ncol=1,                 
    title="District"
)
plt.xlabel('State')
plt.ylabel('Population')
plt.tight_layout()
plt.show()
```

```python
print("ATV for each state:")
avg_atv = state_txnSplit.groupby('State')['ATV (INR)'].mean().reset_index()
avg_atv = avg_atv.sort_values(by='ATV (INR)',ascending=False)
print(avg_atv)
```

```python
# Top 5 States with Highest ATV:
print("Top 5 States with Highest ATV:")
print(avg_atv.head(5))

# Top 5 states with lowest ATV
print("\nTop 5 States with Lowest ATV:")
print(avg_atv.tail(5))
```

```python
print("Total no of App Opens by State, Year and Quarter:")
app_opens = dist_txn.groupby(['State','Year','Quarter'])['App Opens'].sum().reset_index()
print(app_opens.to_string(index=False))
```

```python
selected_state = 'Chandigarh'  
state_data = app_opens[app_opens['State'] == selected_state].copy()
state_data['Time'] = state_data['Year'].astype(str) + ' Q' + state_data['Quarter'].astype(str)
state_data = state_data.sort_values(by=['Year', 'Quarter'])
plt.figure(figsize=(12, 6),dpi=100)
sns.lineplot(data=state_data, x='Time', y='App Opens', marker='o', linewidth=2, color='#1f77b4')
plt.title(f'App Opens Over Time in {selected_state}', fontsize=14, fontweight='bold')
plt.xlabel('Time (Year & Quarter)')
plt.ylabel('App Opens')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
latest_yr = state_txnSplit['Year'].max()
latest_Q = state_txnSplit[state_txnSplit['Year'] == latest_yr]['Quarter'].max()
recent_data = state_txnSplit[(state_txnSplit['Year']==latest_yr) & (state_txnSplit['Quarter']==latest_Q)]
state_txns = recent_data.groupby(['State', 'Transaction Type'])['Transactions'].sum().reset_index()
plt.figure(figsize=(16, 10))
plts = sns.barplot(data=state_txns, x='State', y='Transactions', hue='Transaction Type', palette="Paired")
plt.title(f'Distribution of Transaction Types by State (Q{latest_Q}, {latest_yr})', fontsize=14, fontweight='bold')
plt.xlabel('State')
plt.ylabel('Number of Transactions')
plt.xticks(rotation=90)
plt.legend(title='Transaction Type')
plt.tight_layout()
plt.show()
```

```python
print("Unique District-Code Mapping:")
unique_map = dist_dem[['District','Code']].drop_duplicates().reset_index(drop=True)
print(unique_map.to_string())
unique_map.to_csv('unique_district_code_mapping.csv', index=False)
```

```python
print("Discrepancies Between State and District Level Data:")
district_agg = dist_txn.groupby(['State', 'Year', 'Quarter'])[['Transactions', 'Amount (INR)', 'Registered Users']].sum().reset_index()
comparison = pd.merge(state_txn, district_agg, on=['State', 'Year', 'Quarter'], suffixes=('_State', '_District'))
comparison['Transactions_Match'] = comparison['Transactions_State'] == comparison['Transactions_District']
comparison['Amount_Match'] = comparison['Amount (INR)_State'] == comparison['Amount (INR)_District']
comparison['Users_Match'] = comparison['Registered Users_State'] == comparison['Registered Users_District']
discrepancies= comparison[~(comparison['Transactions_Match'] & comparison['Amount_Match'] & comparison['Users_Match'])]
print(discrepancies[['State', 'Year', 'Quarter', 'Transactions_State', 'Transactions_District',
                  'Amount (INR)_State', 'Amount (INR)_District',
                  'Registered Users_State', 'Registered Users_District']])
```

```python
print("Ratio of Registered Users to Population by State:")
state_users = state_txn.sort_values(by=['Year', 'Quarter']).groupby('State').tail(1)[['State', 'Registered Users']]
state_population = dist_dem.groupby('State')['Population'].sum().reset_index()
merged = pd.merge(state_users, state_population, on='State')
merged['User to Population Ratio'] = merged['Registered Users'] / merged['Population']
print(merged[['State', 'Registered Users', 'Population', 'User to Population Ratio']])
```

```python
plt.figure(figsize=(14, 8))
sns.barplot(data=merged.sort_values('User to Population Ratio', ascending=False),
            x='State', y='User to Population Ratio', hue='State', legend= False, palette='Paired')
plt.xticks(rotation=90)
plt.title('Ratio of Registered Users to Population by State', fontsize=14, fontweight='bold')
plt.xlabel('State')
plt.ylabel('User to Population Ratio')
plt.tight_layout()
plt.show()
```

```python
print("Population Density by Transaction Volume:")
combined = pd.merge(dist_txn, dist_dem, on=['State', 'District'], how='inner')
txn_by_district = combined.groupby(['State', 'District'])[['Transactions', 'Density']].sum().reset_index()
print(txn_by_district)
```

```python
correlation = txn_by_district['Density'].corr(txn_by_district['Transactions'])
print(f"Correlation between Population Density and Transaction Volume: {correlation:.4f}")
```

```python
plt.figure(figsize=(10,8))
sns.scatterplot(data=txn_by_district, x='Density', y='Transactions', hue='State', palette='Paired')
plt.title('Correlation between Population Density and Transaction Volume', fontsize=11, fontweight='bold')
plt.legend(
    loc='center left',       
    bbox_to_anchor=(1.05, 0.5),
    ncol=1,                 
    title="State"
)
plt.xlabel('Population Density')
plt.ylabel('Transaction Volume')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
print("Average Transaction Amount per User by State:")
df = state_txn.sort_values(by=['Year', 'Quarter']).groupby('State').tail(1)
df['Avg Amount per User'] = df['Amount (INR)'] / df['Registered Users']
print(df[['State', 'Amount (INR)', 'Registered Users', 'Avg Amount per User']])
```

```python
# Top 5 highest
top_5 = df.sort_values(by='Avg Amount per User', ascending=False).head(5)
print("Top 5 States with Highest Avg Transaction Amount per User:")
print(top_5[['State', 'Avg Amount per User']])
```

```python
# Top 5 lowest
bottom_5 = df.sort_values(by='Avg Amount per User', ascending=True).head(5)
print("Top 5 States with Lowest Avg Transaction Amount per User:")
print(bottom_5[['State', 'Avg Amount per User']])
```

```python
print("Device Brand Usage Data:")
merged_data = pd.merge(state_dev_data,state_txn,on=['State','Year','Quarter'],suffixes=['_Brand','_Total'])
print(merged_data)
```

```python
print("Usage Ratio Table:")
merged_data['Usage Ratio'] = merged_data['Registered Users_Brand'] / merged_data['Registered Users_Total']
merged_data['Usage Ratio (%)'] = (merged_data['Usage Ratio'] * 100).round(2)
usage_table = merged_data[['State', 'Year', 'Quarter', 'Brand', 'Registered Users_Brand', 'Registered Users_Total', 'Usage Ratio (%)']]
print(usage_table.head(10))  
```

```python
data = merged_data.groupby(['State', 'Brand'])['Usage Ratio'].mean().reset_index()
plt.figure(figsize=(14, 8))
for state in data['State'].unique():
    state_data = data[data['State'] == state]
    plt.bar(state_data['Brand'], state_data['Usage Ratio'] * 100, label=state)

plt.title('Device Brand Usage Ratio by State', fontsize=14, fontweight='bold')
plt.xlabel('Device Brand')
plt.ylabel('Usage Ratio (%)')
plt.xticks(rotation=45)
plt.legend(
    loc='center left',       
    bbox_to_anchor=(1.05, 0.5),
    ncol=1,                 
    title="State"
)
plt.tight_layout()
plt.show()
```

```python
density_corr = correlation_matrix.loc['Density', 'Transactions']
if density_corr > 0.5:
    summary = "There is a strong positive correlation between population density and transaction volume."
elif density_corr > 0.2:
    summary = "There is a moderate positive correlation between population density and transaction volume."
elif density_corr < -0.2:
    summary = "There is a moderate negative correlation between population density and transaction volume."
else:
    summary = "There is little to no correlation between population density and transaction volume."

print("Summary:",summary)
```

# Key Findings and Insights

- Steady Growth: From 2018 to 2021, the number and value of PhonePe transactions have steadily increased in almost all Indian states.

- Top Performing States: The states with the highest transaction volumes and amounts were Uttar Pradesh, Karnataka, and Maharashtra.

- Transaction Categories:

The majority of transactions are Peer-to-Peer (P2P) payments.

Strong growth is also seen in merchant payments and recharge and bill payments, indicating increasing user trust on digital payments.

- Average Transaction Value (ATV): The average value per transaction is steadily rising, showing users' confidence in using PhonePe for larger payments.

- User Base Growth: Particularly in Tier 2 and Tier 3 states, the number of registered users and app opens is increasing rapidly.

- Seasonal Trends: During festive season quarters (Q3 and Q4), spikes are seen, which is indicative of higher digital spending during those times.

# Interpretation (What This Means)

- Digital payment adoption is steadily increasing across India, not just in metropolitan areas.

- PhonePe is not only used for small recharges but also for higher-value payments and peer transfers.

- The expanding user base implies that UPI-based platforms are becoming more popular among a wide range of demographics.

# Actionable Recommendations

- Increase promotional offerings and merchant onboarding in emerging areas such as Bihar, Odisha, and Madhya Pradesh to capitalise on unexplored opportunities.

- To boost PhonePe acceptance, strengthen the merchant ecosystem by offering loyalty rewards and discounts to small retailers.

- Optimize User Experience: To retain new users, increase app speed and simplify the payment transaction.

- Seasonal Campaigns: Launch targeted offers ahead of festival seasons to increase Q3-Q4 transaction volume.

- Data-Driven Insights: Continue to monitor transaction patterns quarterly to inform future market expansion initiatives.
